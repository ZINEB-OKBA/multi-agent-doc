```python
import base64
import json
from PIL import Image
import io
```

```python
# 1. Simuler ou coller ton JSON de réponse FastAPI
output_json ={
"charts": [
    {
      "title": "Occupation mensuelle",
      "type": "bar",
      "base64"
 
}]
}

# 2. Extraire la chaîne base64
raw_base64 = output_json["charts"][0]["base64"]

# 3. Nettoyer le préfixe HTML si présent
if "," in raw_base64:
    raw_base64 = raw_base64.split(",")[1]

# 4. Décoder les octets et afficher l'image
image_bytes = base64.b64decode(raw_base64)
image = Image.open(io.BytesIO(image_bytes))
image.show()  # Ouvre le graphique instantanément sur ton écran !
```
